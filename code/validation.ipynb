#VALIDATION

## data source vs bronze

```python
from bronze_config import ingestion_configuration
```

```python
for table in ingestion_configuration:

    #extract csv file
    source_df = (
        spark.read
        .option("header", "true")
        .option("inferschema", "true")
        .csv(table["file_path"])
    )

    bronze_df = spark.sql(
        f"""
        select
        count(*) as row_count
        from bronze.{table["target_table"]}
        """
    )
    
    source_rows = source_df.count()
    bronze_rows = bronze_df.first()["row_count"]
    
    if source_rows == bronze_rows:
        print(f"Table name: {table["target_table"]} | Status: Matched")
    else:
        print(f"Table name: {table["target_table"]} | Status: Not Matched")
```
